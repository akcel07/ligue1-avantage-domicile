```python
import pandas as pd
df=pd.read_csv("season-2526.csv")
print(df.shape)
print(df.columns)
df.head()
```

```python
import matplotlib.pyplot as plt

df.info()                         # types de colonnes et valeurs manquantes
print('----------------------------------------------------------------------')
print(df.isna().sum())            # nombre de valeurs vides par colonne

#df = df.drop(columns=["Referee"])        # on supprime la colonne vide
df["Date"] = pd.to_datetime(df["Date"])  # on transforme le texte en vraie date
print('----------------------------------------------------------------------')
print(df.duplicated().sum()) #on vérifie si on a pas la meme ligne qui se repete 
print(df.duplicated(subset=["Date", "HomeTeam", "AwayTeam"]).sum()) #pas le meme match
```

```python
print(df[["FTHG", "FTAG"]].describe())
print(df[df["FTAG"] == 7])
print('----------------------------------------------------------------------')
print(df["FTR"].value_counts())                        # nombre de H, D, A
pourcentages = df["FTR"].value_counts(normalize=True) * 100
print(pourcentages.round(1))

pourcentages.plot(kind="bar")
plt.title("Résultats des matchs de Ligue 1 (2025/26)")
plt.ylabel("% des matchs")
plt.show()

print(df["FTHG"].mean(), df["FTAG"].mean())           # buts moyens à domicile / extérieur
```

```python
saisons = ["season-2122", "season-2223", "season-2324", "season-2425", "season-2526"]

tableaux = []
for nom in saisons:
    t = pd.read_csv(nom + ".csv")                 # 1. lire le fichier
    t["Saison"] = nom[-4:]                        # 2. ajouter la colonne (ex. "2526")
    tableaux.append(t[["Saison", "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR"]])
    
df_all = pd.concat(tableaux, ignore_index=True)   # 3. tout empiler
print('----------------------------------------------------------------------')
df_all.shape
df_all["Saison"].value_counts()
print('----------------------------------------------------------------------')
resultats = pd.crosstab(df_all["Saison"], df_all["FTR"], normalize="index") * 100
resultats.round(1)
```

```python
resultats.plot(kind="bar", figsize=(10, 6))
plt.title("Résultats des matchs de Ligue 1 par saison")
plt.xlabel("Saison")
plt.ylabel("% des matchs")
plt.xticks(rotation=0)
plt.legend(title="Résultat (H = domicile, D = nul, A = extérieur)")
plt.show()
```
